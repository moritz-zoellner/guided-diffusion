```python
import h5py
```

```python
path = "../data/can/mh/low_dim_v15.hdf5"

def print_tree(name, obj):
    if isinstance(obj, h5py.Dataset):
        print(f"DATASET: {name} shape={obj.shape}")
    else:
        print(f"GROUP:   {name}")

with h5py.File(path, "r") as f:
    f.visititems(print_tree)
```
